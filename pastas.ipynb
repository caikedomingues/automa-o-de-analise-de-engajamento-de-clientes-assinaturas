Esse arquivo será necessário para automatozar a criação das pastas necessárias para o sistema

Import da biblioteca os

```python
# Tem como objetivo conectar o python ao sistema de arquivos do sistema 
# operacional
import os

# Permite manipular datas e horas do sistena
import datetime
```

Criando as pastas necessárias

```python
# Ira inspecionar o bloco de código com o objetivo de capturar possiveis 
# erros de execuçao.
try:
    # makedirs: Função da biblioteca os que tem como objetivo criar diretórios
    # no sistema. Ela recebe como argumento o caminho de criação da pasta e
    # um comando que evita que o código seja quebrado caso a pasta ja tenha sido
    # criada.     
    os.makedirs("Treinamento", exist_ok=True)
    os.makedirs("Dados", exist_ok=True)
    os.makedirs("Logs", exist_ok=True)
    os.makedirs("Relatorios", exist_ok=True)
    os.makedirs("Historico", exist_ok=True)

    # Irá coletar a data do sistema com o objetivo de diferenciar os arquivos
    # de logs. Dessa forma, poderemos ter um arquivo log por dia.
    data_atual = datetime.datetime.now().strftime("%d-%m-%Y")

    # Ira coletar a hora do sistema com o objetivo de registrar o horário
    # de execução de cada etapa. 
    hora = datetime.datetime.now().strftime("%H-%M")

    # Ira criar e editar os arquivos logs que indicam a criação das pastas.
    # open: Função do python que tem como objetivo criar e carregar arquivos
    # no código python. Basicamente a função irá criar o arquivo e atualiza-lo
    # futuramente com novas informações, ele recebe como argumento:

    # f"Logs/logs_{data_atual}.txt": Caminho que o arquivo ficara armazenado
    # após ser criado.

    # "a": Comando "append" que adiciona dados nos arquivos sem sobrescreve-los.
    # Observação: Caso o argumento passado seja "w"(write), o python irá sobrescrever
    # os dados que foram escritos anteriormente. 

    # encoding: Adapta o Python a entender caracteres especiais como acentos
    # e pontuações.

    # log: Apelido que representa o arquivo que está sendo criado/modificado.
    with open(f"Logs/logs_{data_atual}.txt", "a", encoding="utf-8") as log:

        # write: Função que tem como objetivo registrar dados nos arquivos.
        log.write("Os arquivos foram criados com sucesso\n")
        log.write(f"Hora da execução: {hora}\n")
        log.write("------------------------------------------------------------------------------------------------------------------------------\n")

# Exceção que tem como objetivo capturar falhas na criação das pastas do projeto 
except OSError as erro:

    # Irá repetir o processo de coleta de data 
     data_atual = datetime.datetime.now().strftime("%d-%m-%Y")

    # Irá repetir o processo de coleta de horário
     hora = datetime.datetime.now().strftime("%H-%M")

    # Ira repetir a criação/modificação do log do sistema
     with open(f"Logs/logs_{data_atual}.txt", "a", encoding="utf-8") as log:

        log.write(f"Falha na criação\atualização das pastas: {erro}\n")
        log.write(f"Hora da execução: {hora}\n")
        log.write("------------------------------------------------------------------------------------------------------------------------------\n")


    

```
